**DecisionLens AI**

An AI-powered decision support platform.

Problem: Making complex decisions can be difficult when multiple alternatives, evaluation criteria, and potential risks must be considered.

Solution: DecisionLens AI provides a structured decision-support system that evaluates alternatives, compares weighted criteria, identifies potential risks, and generates explainable recommendations.

Core Technologies: Python, Pandas, Scikit-learn, NLP, Plotly, Streamlit.

```python
criteria = {
    "Cost": 0.30,
    "Security": 0.30,
    "Scalability": 0.25,
    "Ease of Use": 0.15
}

print("Decision Criteria:")
for criterion, weight in criteria.items():
    print(f"{criterion}: {weight * 100}%")

print("Total Weight:", sum(criteria.values()) * 100, "%")
```

Add the alternatives:i give each alternative a score from 1 to 10 against each criterion.

```python
alternatives = {
    "AWS": {
        "Cost": 7,
        "Security": 9,
        "Scalability": 10,
        "Ease of Use": 7
    },

    "Azure": {
        "Cost": 8,
        "Security": 9,
        "Scalability": 9,
        "Ease of Use": 8
    },

    "Google Cloud": {
        "Cost": 8,
        "Security": 8,
        "Scalability": 9,
        "Ease of Use": 9
    }
}

scores = pd.DataFrame(alternatives).T

scores
```

Build the Decision Engine: A weighted decision-making algorithm.

It calculates:

S = sum (Score_i * Weight_i)


```python
def calculate_scores(alternatives, criteria):

    if not np.isclose(sum(criteria.values()), 1):
        raise ValueError("Criteria weights must add up to 1.")

    results = {}

    for alternative, scores in alternatives.items():

        total_score = 0

        for criterion, weight in criteria.items():

            score = scores[criterion]
            total_score += score * weight

        results[alternative] = round(total_score, 2)

    return results


results = calculate_scores(alternatives, criteria)

print("Decision Results:")

for name, score in sorted(
    results.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(name, ":", score)
```

```python
results_df = pd.DataFrame(
    list(results.items()),
    columns=["Alternative", "Final Score"]
)

fig = px.bar(
    results_df,
    x="Alternative",
    y="Final Score",
    title="DecisionLens AI - Alternative Ranking",
    text="Final Score"
)

fig.update_traces(textposition="outside")
fig.update_layout(
    yaxis_range=[0, 10],
    template="plotly_white"
)

fig.show()
```

```python
new_criteria = {
    "Cost": 0.15,
    "Security": 0.45,
    "Scalability": 0.25,
    "Ease of Use": 0.15
}

new_results = calculate_scores(
    alternatives,
    new_criteria
)

print("Original Ranking:", results)
print("New Ranking:", new_results)
```

Add the user's decision problem

```python
decision_problem = """
We need a cloud platform for a new software project.
Security and scalability are important.
The platform should also provide reasonable costs
and be easy for developers to use.
"""
```

Let NLP analyze the descriptions

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np

# Prepare the documents
documents = [
    decision_problem
] + [
    alternative_descriptions[name]
    for name in alternative_descriptions
]

# Create TF-IDF model
vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2)
)

# Convert text into numerical features
tfidf_matrix = vectorizer.fit_transform(documents)

# Calculate similarity
similarities = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:]
)[0]

# Store results
nlp_results = {}

for name, similarity in zip(
    alternative_descriptions.keys(),
    similarities
):
    nlp_results[name] = round(float(similarity) * 100, 2)

print("AI Relevance Analysis:")

for name, score in nlp_results.items():
    print(f"{name}: {score}%")
```

```python
nlp_df = pd.DataFrame(
    list(nlp_results.items()),
    columns=["Alternative", "Relevance Score"]
)

fig = px.bar(
    nlp_df,
    x="Alternative",
    y="Relevance Score",
    title="DecisionLens AI - NLP Relevance Analysis",
    text="Relevance Score"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    template="plotly_white",
    yaxis_title="Text Relevance (%)"
)

fig.show()
```

```python
final_results = {}

for name in results:

    decision_score = results[name]
    relevance_score = nlp_results[name] / 10

    final_score = (
        0.7 * decision_score +
        0.3 * relevance_score
    )

    final_results[name] = round(final_score, 2)

ranking = pd.DataFrame(
    list(final_results.items()),
    columns=["Alternative", "Final Score"]
)

ranking = ranking.sort_values(
    by="Final Score",
    ascending=False
).reset_index(drop=True)

ranking
```


The system should not only tell users which option scored highest. It should also warn them about potential risks.

For example:
- High cost → Budget risk.
- Security weaknesses → Security risk.
- Complex implementation → Operational risk.
- Limited scalability → Growth risk.

```python
risk_keywords = {
    "Financial Risk": [
        "expensive", "high cost", "budget",
        "pricing", "costly"
    ],

    "Security Risk": [
        "vulnerability", "weak security",
        "data breach", "security issue"
    ],

    "Scalability Risk": [
        "limited scalability", "capacity",
        "performance limitation"
    ],

    "Operational Risk": [
        "complex", "difficult to manage",
        "maintenance", "technical complexity"
    ],

    "Dependency Risk": [
        "vendor lock-in", "vendor dependency",
        "proprietary"
    ]
}
```

```python
def detect_risks(description):

    description = description.lower()
    detected_risks = []

    for risk, keywords in risk_keywords.items():

        matched_keywords = [
            word for word in keywords
            if word in description
        ]

        if matched_keywords:
            detected_risks.append({
                "Risk": risk,
                "Evidence": matched_keywords,
                "Description": (
                    f"Potential {risk.lower()} detected "
                    f"because the description mentions "
                    f"{', '.join(matched_keywords)}."
                )
            })

    return detected_risks
```

```python
risk_results = {}

for name, description in alternative_descriptions.items():

    risk_results[name] = detect_risks(description)

    print(f"\n{name}")
    print("-" * 30)

    if risk_results[name]:

        for risk in risk_results[name]:
            print("Risk:", risk["Risk"])
            print("Evidence:", risk["Evidence"])

    else:
        print("No keyword-based risks detected.")
```

Give every alternative a risk level.

For this prototype, we'll use the number of detected risk categories as an initial indicator. This is not a measure of actual probability or severity.

```python
def calculate_risk_level(risks):

    risk_count = len(risks)

    if risk_count == 0:
        return "No flagged risks"

    elif risk_count == 1:
        return "Low"

    elif risk_count == 2:
        return "Medium"

    else:
        return "High"


risk_summary = []

for name, risks in risk_results.items():

    risk_summary.append({
        "Alternative": name,
        "Detected Risks": len(risks),
        "Risk Level": calculate_risk_level(risks)
    })

risk_df = pd.DataFrame(risk_summary)

display(risk_df)
```

Explainable Recommendations
Instead of simply saying AWS is the winner, the application should explain why.

```python
def generate_recommendation(name):

    score = results[name]
    relevance = nlp_results[name]
    risks = risk_results[name]

    explanation = []

    explanation.append(
        f"{name} achieved a weighted decision score "
        f"of {score}/10."
    )

    explanation.append(
        f"Its text relevance score is {relevance}%."
    )

    if risks:
        explanation.append(
            f"{len(risks)} potential risk categories "
            "were flagged for further investigation."
        )

        for risk in risks:
            explanation.append(
                f"{risk['Risk']}: "
                f"{risk['Description']}"
            )
    else:
        explanation.append(
            "No risks were flagged by the current "
            "keyword-based detector."
        )

    return explanation


for name in results:

    print("\n", name)
    print("=" * 40)

    for point in generate_recommendation(name):
        print("-", point)
```

```python
comparison_report = ranking.copy()

comparison_report["NLP Relevance"] = [
    nlp_results[name]
    for name in comparison_report["Alternative"]
]

comparison_report["Risk Count"] = [
    len(risk_results[name])
    for name in comparison_report["Alternative"]
]

comparison_report["Risk Level"] = [
    calculate_risk_level(risk_results[name])
    for name in comparison_report["Alternative"]
]

display(comparison_report)
```

part 2 : Create the application interface

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import plotly.express as px

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

st.set_page_config(
    page_title="DecisionLens AI",
    page_icon="🔎",
    layout="wide"
)

st.title("🔎 DecisionLens AI")
st.caption("Think clearly. Decide confidently.")

st.markdown(
    "An explainable decision-support tool "
    "for comparing alternatives and identifying risks."
)

st.header("1. Define Your Decision")

decision = st.text_area(
    "Describe the decision you need to make",
    placeholder="Example: Choose a cloud platform for a new software project."
)

st.header("2. Add Your Alternatives")

num_options = st.number_input(
    "Number of alternatives",
    min_value=2,
    max_value=5,
    value=3
)

alternatives = {}

for i in range(int(num_options)):

    with st.expander(f"Alternative {i+1}", expanded=True):

        name = st.text_input(
            "Alternative name",
            value=["AWS", "Azure", "Google Cloud", "Option 4", "Option 5"][i],
            key=f"name_{i}"
        )

        description = st.text_area(
            "Description",
            key=f"description_{i}"
        )

        alternatives[name] = description

st.header("3. Define Evaluation Criteria")

criteria_names = [
    "Cost",
    "Security",
    "Scalability",
    "Ease of Use"
]

weights = {}

cols = st.columns(4)

for i, criterion in enumerate(criteria_names):

    with cols[i]:

        weights[criterion] = st.number_input(
            criterion + " weight (%)",
            min_value=0,
            max_value=100,
            value=[30, 30, 25, 15][i],
            key=f"weight_{i}"
        )

st.info(
    f"Total weight: {sum(weights.values())}% "
    "(must equal 100%)"
)

st.header("4. Score Each Alternative")

scores = {}

for name in alternatives:

    scores[name] = {}

    with st.expander(f"Scores for {name}"):

        cols = st.columns(4)

        for i, criterion in enumerate(criteria_names):

            with cols[i]:

                scores[name][criterion] = st.slider(
                    criterion,
                    min_value=1,
                    max_value=10,
                    value=7,
                    key=f"score_{name}_{criterion}"
                )

risk_keywords = {
    "Financial Risk": [
        "expensive", "high cost", "budget",
        "pricing", "costly"
    ],
    "Security Risk": [
        "vulnerability", "weak security",
        "data breach", "security issue"
    ],
    "Scalability Risk": [
        "limited scalability", "capacity",
        "performance limitation"
    ],
    "Operational Risk": [
        "complex", "difficult to manage",
        "maintenance", "technical complexity"
    ],
    "Dependency Risk": [
        "vendor lock-in", "vendor dependency",
        "proprietary"
    ]
}

if st.button("Analyze Decision", type="primary"):

    if not decision.strip():
        st.error("Please describe your decision.")

    elif any(not n.strip() for n in alternatives):
        st.error("Please enter a name for every alternative.")

    elif any(not d.strip() for d in alternatives.values()):
        st.error("Please describe every alternative.")

    elif len(set(alternatives.keys())) != len(alternatives):
        st.error("Alternative names must be unique.")

    elif sum(weights.values()) != 100:
        st.error("Evaluation weights must total 100%.")

    else:

        documents = [decision] + list(alternatives.values())

        vectorizer = TfidfVectorizer(
            stop_words="english",
            ngram_range=(1, 2)
        )

        matrix = vectorizer.fit_transform(documents)

        similarities = cosine_similarity(
            matrix[0:1],
            matrix[1:]
        )[0]

        results = []

        for (name, description), similarity in zip(
            alternatives.items(), similarities
        ):

            decision_score = sum(
                scores[name][criterion] *
                weights[criterion] / 100
                for criterion in criteria_names
            )

            relevance = float(similarity) * 100

            final_score = (
                0.7 * decision_score +
                0.3 * (relevance / 10)
            )

            description_lower = description.lower()

            risks = {
                risk: [
                    word for word in keywords
                    if word in description_lower
                ]
                for risk, keywords in risk_keywords.items()
            }

            risks = {
                risk: evidence
                for risk, evidence in risks.items()
                if evidence
            }

            results.append({
                "Alternative": name,
                "Decision Score": round(decision_score, 2),
                "NLP Relevance": round(relevance, 2),
                "Final Score": round(final_score, 2),
                "Detected Risks": len(risks),
                "Risk Details": risks
            })

        report = pd.DataFrame(results)

        report = report.sort_values(
            "Final Score",
            ascending=False
        ).reset_index(drop=True)

        st.success("Decision analysis completed!")

        st.header("5. Analysis Results")

        best = report.iloc[0]

        st.metric(
            "Top Ranked Alternative",
            best["Alternative"],
            f"{best['Final Score']}/10"
        )

        st.subheader("Alternative Ranking")

        st.dataframe(
            report[
                [
                    "Alternative",
                    "Decision Score",
                    "NLP Relevance",
                    "Final Score",
                    "Detected Risks"
                ]
            ],
            use_container_width=True
        )

        fig = px.bar(
            report,
            x="Alternative",
            y="Final Score",
            color="Final Score",
            text="Final Score",
            title="Final Alternative Ranking"
        )

        fig.update_layout(template="plotly_white")

        st.plotly_chart(
            fig,
            use_container_width=True
        )

        st.header("6. Risk Analysis")

        for _, row in report.iterrows():

            with st.expander(
                f"{row['Alternative']} — "
                f"{row['Detected Risks']} flagged risks"
            ):

                if row["Risk Details"]:

                    for risk, evidence in row["Risk Details"].items():

                        st.warning(
                            f"{risk}: Evidence found: "
                            f"{', '.join(evidence)}"
                        )

                else:
                    st.success(
                        "No keyword-based risk signals detected."
                    )

        st.header("7. Explainable Recommendation")

        st.write(
            f"The highest-ranked alternative is "
            f"**{best['Alternative']}**, based on the "
            f"criteria and weights you supplied."
        )

        st.write(
            "This is a ranking of the provided inputs, "
            "not a guarantee that this alternative is objectively best."
        )

        st.download_button(
            "Download Decision Report",
            report.drop(columns=["Risk Details"]).to_csv(index=False),
            file_name="decisionlens_report.csv",
            mime="text/csv"
        )
```
